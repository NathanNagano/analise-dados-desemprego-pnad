# Análise de Dados Públicos – Desemprego no Brasil

Este projeto tem como objetivo analisar a evolução da taxa de desemprego no Brasil
utilizando dados públicos da PNAD Contínua (IBGE).

A análise considera dados trimestrais recentes, buscando identificar tendências
e padrões no mercado de trabalho brasileiro.


```python
# Importando as bibliotecas
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("desemprego.csv")
df.head()
```

## Entendimento do Dataset

O conjunto de dados possui as seguintes colunas:

- **periodo**: trimestre de referência (ex: 1º tri 2024).
- **valor**: taxa de desemprego percentual no período.

```python
# Renomeando a coluna para melhor entendimento
df.rename(columns={"valor": "taxa_desemprego_percentual"}, inplace=True)

# Informações gerais
df.info()
```

```python
# Descrição das estatísticas
df["taxa_desemprego_percentual"].describe()
```

```python
# Análise de Extremos
maior = df.loc[df["taxa_desemprego_percentual"].idxmax()]
menor = df.loc[df["taxa_desemprego_percentual"].idxmin()]

print("Maior taxa de desemprego:")
print(maior)

print("\nMenor taxa de desemprego:")
print(menor)
```

```python
# Visualização de Dados
plt.figure()
plt.plot(df["periodo"], df["taxa_desemprego_percentual"])
plt.xlabel("Período")
plt.ylabel("Taxa de Desemprego (%)")
plt.title("Evolução da Taxa de Desemprego no Brasil")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

## Conclusões

- Observa-se uma tendência geral de queda da taxa de desemprego ao longo do período analisado.
- O maior índice ocorre no início de 2023, seguido por redução progressiva.
- O uso de dados trimestrais permite identificar oscilações no mercado de trabalho.
- Dados públicos são fundamentais para análises econômicas e sociais.
